```python
# visualize_anomalies.ipynb

import json
import pandas as pd
import matplotlib.pyplot as plt

# Load JSON output
with open("../models/anomaly_output_standard.json") as f:
    data = json.load(f)

# Convert to DataFrame
df = pd.DataFrame(data)

# Sort by anomaly_score (descending)
df_sorted = df.sort_values("anomaly_score", ascending=False)

# Show top 10 most anomalous transactions
print("Top 10 anomalous transactions:")
display(df_sorted.head(10))

# Plot histogram of anomaly scores
plt.figure(figsize=(8,4))
plt.hist(df["anomaly_score"], bins=50, color='skyblue')
plt.title("Anomaly Scores Distribution")
plt.xlabel("Anomaly Score")
plt.ylabel("Count")
plt.show()

# Optional: Show top features for top 5 anomalies
print("Top features for top 5 anomalies:")
for i, row in df_sorted.head(5).iterrows():
    print(f"Transaction ID: {row['transaction_id']}")
    print("Top Features:", row['top_features'])
    print("SHAP Explanation:", row['explanation'])
    print("-"*50)
```
